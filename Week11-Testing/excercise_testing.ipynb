```python
%%ipytest -v

# Function under test
def is_palindrome(s):
    cleaned = s.lower().replace(" ", "")
    return cleaned == cleaned[::-1]

# Happy-path cases
@pytest.mark.parametrize("s, expected", [
    ("racecar", True),
    ("Madam", True),
    ("hello", False),
    ("Python", False),
])
def test_is_palindrome_returns_correct_result(s, expected):
    assert is_palindrome(s) == expected

# Edge case: empty string
def test_is_palindrome_empty_string_returns_true():
    assert is_palindrome("") is True

# Edge case: spaces are ignored
def test_is_palindrome_ignores_spaces():
    assert is_palindrome("race car") is True

# Edge case: only spaces
def test_is_palindrome_only_spaces_returns_true():
    assert is_palindrome("   ") is True

# Edge case: single character
def test_is_palindrome_single_character_returns_true():
    assert is_palindrome("a") is True
```
